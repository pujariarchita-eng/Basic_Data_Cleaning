```python
import pandas as pd

# ------------------------------------------------------------
# 1. LOAD THE REAL-WORLD DATASET
# ------------------------------------------------------------

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

df = pd.read_csv(url)

print("=" * 60)
print("TITANIC DATA CLEANING PROJECT")
print("=" * 60)

# Save original dataset
df.to_csv("titanic_original.csv", index=False)

print("\n1. ORIGINAL DATASET")
print("-" * 60)
print("Shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())

# ------------------------------------------------------------
# 2. INSPECT DATA TYPES AND MISSING VALUES
# ------------------------------------------------------------

print("\n2. DATA TYPES")
print("-" * 60)
print(df.dtypes)

print("\n3. MISSING VALUES BEFORE CLEANING")
print("-" * 60)
print(df.isnull().sum())

print("\n4. DUPLICATES BEFORE CLEANING")
print("-" * 60)
print("Duplicate rows:", df.duplicated().sum())

print("\n5. ORIGINAL DATASET INFORMATION")
print("-" * 60)
df.info()

# ------------------------------------------------------------
# 3. CREATE A COPY FOR CLEANING
# ------------------------------------------------------------

cleaned_df = df.copy()

# ------------------------------------------------------------
# 4. HANDLE MISSING VALUES
# ------------------------------------------------------------

# Fill numerical missing values with the median
for column in ["age", "fare"]:
    if column in cleaned_df.columns:
        cleaned_df[column] = cleaned_df[column].fillna(
            cleaned_df[column].median()
        )

# Fill categorical missing values with the mode
for column in ["embarked", "embark_town"]:
    if column in cleaned_df.columns:
        if not cleaned_df[column].mode().empty:
            cleaned_df[column] = cleaned_df[column].fillna(
                cleaned_df[column].mode()[0]
            )

# Drop deck because it has a high proportion of missing values
if "deck" in cleaned_df.columns:
    cleaned_df = cleaned_df.drop(columns=["deck"])

print("\n6. MISSING VALUES AFTER CLEANING")
print("-" * 60)
print(cleaned_df.isnull().sum())

# ------------------------------------------------------------
# 5. REMOVE DUPLICATE RECORDS
# ------------------------------------------------------------

duplicates_removed = cleaned_df.duplicated().sum()

cleaned_df = cleaned_df.drop_duplicates().reset_index(drop=True)

print("\n7. DUPLICATE REMOVAL")
print("-" * 60)
print("Duplicate rows removed:", duplicates_removed)
print("Rows after duplicate removal:", len(cleaned_df))

# ------------------------------------------------------------
# 6. CORRECT DATA TYPES
# ------------------------------------------------------------

# Convert integer-like numerical columns to integer types
for column in ["survived", "pclass", "sibsp", "parch"]:
    if column in cleaned_df.columns:
        cleaned_df[column] = cleaned_df[column].astype("int64")

# Convert text columns to Pandas category type
categorical_columns = [
    "sex", "embarked", "class", "who",
    "adult_male", "alive", "alone", "embark_town"
]

for column in categorical_columns:
    if column in cleaned_df.columns:
        cleaned_df[column] = cleaned_df[column].astype("category")

# ------------------------------------------------------------
# 7. SAVE THE CLEANED DATASET
# ------------------------------------------------------------

cleaned_df.to_csv("titanic_cleaned.csv", index=False)

# ------------------------------------------------------------
# 8. DISPLAY FINAL RESULTS
# ------------------------------------------------------------

print("\n8. CLEANED DATASET")
print("=" * 60)

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", cleaned_df.shape)

print("\nFinal data types:")
print(cleaned_df.dtypes)

print("\nMissing values in cleaned dataset:")
print(cleaned_df.isnull().sum())

print("\nRemaining duplicate rows:", cleaned_df.duplicated().sum())

print("\nFirst 5 cleaned records:")
print(cleaned_df.head())

print("\nStatistical summary:")
print(cleaned_df.describe(include="all"))

print("\n9. CLEANING SUMMARY")
print("=" * 60)
print("1. Loaded and inspected the Titanic dataset.")
print("2. Saved the original dataset.")
print("3. Filled missing age and fare values using medians.")
print("4. Filled missing embarkation values using modes.")
print("5. Removed the high-missing-value deck column.")
print("6. Removed duplicate records.")
print("7. Corrected numerical and categorical data types.")
print("8. Exported the cleaned dataset.")

print("\nFiles created successfully:")
print("- titanic_original.csv")
print("- titanic_cleaned.csv")
print("\nData cleaning workflow completed!")
```
